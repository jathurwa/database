```python
#calculating number of HIV cases for each country over the whole period

HIV_Total_cases = dataframe.groupby('Location')['Value(000)'].sum()
print (HIV_Total_cases)
```

```python
#Sorting the total case per country in ascending order
HIV_Total_cases = HIV_Total_cases.sort_values(ascending=False)
print (HIV_Total_cases)
```

```python
Overall_total_cases = HIV_Total_cases.sum()
print (Overall_total_cases)
```

```python
cumulative_percentage_per_country = HIV_Total_cases.cumsum() / HIV_Total_cases.sum() * 100
print(cumulative_percentage_per_country)
```
